```python
#1.find the titanic dataset online or github
#2.find total number of passengers
#3.find the average age of surviors and non surviors
#4.list the passengers who are older than 60 years
#5.find the avg fare for each passenger class

import numpy as np
import pandas as pd

url = pd.read_csv('titanic.csv')
df = pd.DataFrame(url)

#finding the total number of passengers
print("The total Number of passengers : ",df['PassengerId'].count())
print("\n")

#find the average age of surviors and non surviors
df['survived_num'] = df['Survived']==1
print("The Average age of survivors:")
print(df.groupby('survived_num')['Age'].mean())
print("\n")

#without using groupby
print("The average age of non survivors : ")
non_survivors = df[~(df['survived_num'])]
print(non_survivors['Age'].mean())
print("\n")

#list the passengers who are older than 60 years
print("The list of passengers who are older than 60 years of age :")
print(df[(df['Age'] > 60)][['PassengerId','Name', 'Age', 'Sex']])
print("\n")

#find the avg fare for each passenger class
print(df.groupby('Pclass')['Fare'].mean())

```
