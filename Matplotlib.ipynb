```python
from sklearn.datasets import load_iris
iris = load_iris(as_frame=True)
df = iris.frame
df.head()
```

```python
plt.scatter(df['sepal length (cm)'],df['petal length (cm)'])

plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.title("Comparing the length using MatPlotLib library")
plt.show()
```

```python
#Activity 3:
plt.hist(df['petal width (cm)'])
plt.title("Distribution")
plt.show()
```

```python
#Activity 4:
species = ['Setosa','Versicolor', 'Virginica']
count = df['target'].value_counts()
plt.bar(species, count)
plt.show()
```

```python
import pandas as pd
import numpy as np

url = pd.read_csv('Chennai.csv')
df1 = pd.DataFrame(url)

url1 = pd.read_csv('Mumbai.csv')
df2 = pd.DataFrame(url1)

monthdf1 = df1[df1['MONTH']==7]
monthdf2 = df2[df2['MONTH']==7]
temp1 = df1[df1['TEMPERATURE'], monthdf1]
temp2 = df2[df2['TEMPERATURE'], monthdf2]

print(temp1)
print(temp2)


plt.plot(temp1, label = 'Chennai')
plt.plot(temp2, label = 'Mumbai')
plt.title("Temperature in Chennai and Mumbai")
plt.xlabel("Days")
plt.ylabel("Temperature")
plt.show()

#I have not plotted for just a single month instead for an year

print(df1)
print(df2)



```
